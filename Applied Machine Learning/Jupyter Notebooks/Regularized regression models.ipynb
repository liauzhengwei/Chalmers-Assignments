## Regularized regression models (toy example)

We try out the basic least-squares linear regression and some regularized linear regression models: ridge, lasso, elastic net.

```python
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
```

We make some synthetic data: first we generate a 100-dimensional feature matrix with 100,000 instances, all normally distributed. We then form the output variable as a simple linear relation, by taking 0.4 times the first feature and -1.8 times the second feature. We then add a bit of noise.

As you see, the first two features are the only ones that are actually used. Let's see if the various regression models can discover this fact.

```python
X = np.random.normal(size=(100000, 100))
Y = 0.4*X[:,0] - 1.8*X[:,1] + 0.5*np.random.normal(size=100000)
```

The standard least-squares regression model comes fairly close to recovering the true weights (0.4 and -1.8). All the other weights are fairly small, although not strictly zero.

```python
lr = LinearRegression()
lr.fit(X, Y)
lr.coef_
```

The same holds for the ridge regression (least squares plus L2 regularization). If you want, you can play around with the strength of the regularizer (`alpha`) and see how this affects the first two coefficients.

```python
ridge = Ridge(alpha=10000)
ridge.fit(X, Y)
ridge.coef_
```

When we try out the Lasso (least squares plus L1 regularization), we clearly see the effect of the feature selection property of the Lasso: it has set most of the coefficients to exactly zero.

With the default value of `alpha` (which is 1), it seems we get a case of underfitting, because the Lasso has turned off the first feature. If we reduce it a bit, we more or less recover the true weights of the two first features and set all others to zero. (If we reduce it too much, non-zero elements will start to appear.)

```python
lasso = Lasso(alpha=0.01)
lasso.fit(X, Y)
lasso.coef_
```

The Elastic Net model uses a combination of L1 and L2 regularization. It has a feature selection effect similar to the Lasso.

```python
en = ElasticNet(alpha=0.01, l1_ratio=0.5)

en.fit(X, Y)
en.coef_
```
